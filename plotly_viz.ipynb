## Feature Histogram
* Dropdown: Select attributes
* Slider: Number of bins

```python
def feat_histogram(df: pd.DataFrame, feature: str, nbins: int):
    fig = px.histogram(df, x=feature, nbins=nbins)
    return fig
```

## Distribution 
* Tick box: Species(All, Adelie, Chinstrap, Gentoo)

```python
def bar_species_island(df: pd.DataFrame):
    # Color theme
    color_map = {
        "Adelie": "darkorange",
        "Chinstrap": "purple",
        "Gentoo": "#008B8B"
    }

    count_df = penguins_df.groupby("island")["species"].value_counts().reset_index()
    count_df= count_df.sort_values(by="species")
    fig = px.bar(
        count_df, x="island", y="count", color="species",
        color_discrete_map = color_map,
        title="Distribution of Species Among Islands", 
        subtitle="While Adele spreads across three islands, Chinstrap and Gentoo sticks to Dream and Biscoe respectively"
    )
    return fig
```

```python
bar_species_island(penguins_df)
```

## Correlation 
* Tick box: Species in scatterplot (All, Adelie, Chinstrap, Gentoo)
* Tick box: OLS (Yes, No)

```python
def corr_flipper_body(df: pd.DataFrame):
    # Color theme
    color_map = {
        "Adelie": "darkorange",
        "Chinstrap": "purple",
        "Gentoo": "#008B8B"
    }
    
    fig = px.scatter(
        penguins_df,
        x="flipper_length_mm", y="body_mass_g",
        color="species", facet_col="island",
        color_discrete_map=color_map,
        trendline="ols",
        hover_data=['species', 'island', 'sex'],
        title="Flipper Length and Body Mass Per Island",
        # morphology is driven more by species than geography
        subtitle="All species show a positive relationship, " \
        "with Adelie remaining consistent across islands, " \
        "indicating morphology is driven more by species than geography."
    )

    fig.update_layout(template="plotly_white")
    return fig
```

```python
corr_flipper_body(penguins_df)
```

```python
def corr_bill_body(df: pd.DataFrame):
    # Color theme
    color_map = {
        "Adelie": "darkorange",
        "Chinstrap": "purple",
        "Gentoo": "#008B8B"
    }
    
    fig = px.scatter(
        penguins_df,
        x="bill_area_mm2", y="body_mass_g",
        color="species", facet_col="island",
        color_discrete_map=color_map,
        trendline="ols",
        hover_data=['species', 'island', 'sex'],
        title="Bill Area and Body Mass Per Island",
        # morphology is driven more by species than geography
        subtitle="All species show a positive relationship, " \
        "with Adelie remaining consistent across islands, " \
        "indicating morphology is driven more by species than geography."
    )

    fig.update_layout(template="plotly_white")
    return fig
```

```python
corr_bill_body(penguins_df)
```

## How big is Body Mass to Bill Area

```python
def box_body_bill(df):
    # Color theme
    color_map = {
        "Adelie": "darkorange",
        "Chinstrap": "purple",
        "Gentoo": "#008B8B"
    }
    
    df["mass_per_bill_area"] = (penguins_df["body_mass_g"] / (penguins_df["bill_length_mm"] * penguins_df["bill_depth_mm"]))

    fig = px.box(
        penguins_df,
        x="species",
        y="mass_per_bill_area",
        color="species",
        color_discrete_map={
        "Adelie": "darkorange",
        "Chinstrap": "purple",
        "Gentoo": "#008B8B"
        },
        title="Body Mass per Unit Bill Area by Species"
    )

    fig.update_layout(
    template="plotly_white",
    yaxis_title="Mass per Bill Area (g / mm²)"
    )

    return fig
```

```python
box_body_bill(penguins_df)
```

## Ground truth labels

```python
def plot_pca_clusters(df, sil_score, ari_score):
    import plotly.express as px

    fig = px.scatter(
        df,
        x='PC1',
        y='PC2',
        color='cluster',
        hover_data=['species', 'island', 'sex'],
        title=f"PCA Clustering | Silhouette: {sil_score:.2f} | ARI: {ari_score:.2f}"
    )

    return fig
plot_pca_clusters(penguins_df, sil_score, ari)
```

```python
def plot_pca_clusters(df):
    fig = px.scatter(
        df,
        x='PC1',
        y='PC2',
        color='cluster',
        hover_data=['species', 'island', 'sex'],
    )
    return fig

plot_pca_clusters(penguins_df)
```

```python
def plot_pca_species(df):
    fig = px.scatter(
        df,
        x='PC1',
        y='PC2',
        color='species',
        title="Ground Truth (Species)"
    )

    return fig
plot_pca_species(penguins_df)
```

```python
cluster_map = {
    0: "Adelie",
    1: "Gentoo",
    2: "Chinstrap"
}

penguins_df["cluster_label"] = penguins_df["cluster"].map(cluster_map)
penguins_df["is_mismatch"] = penguins_df["cluster_label"] != penguins_df["species"]
```

```python
penguins_df
```

```python
def plot_pca_clusters(df):
    fig = px.scatter(
        df,
        x='PC1',
        y='PC2',
        color='cluster_label',
        symbol='is_mismatch',  # 👈 key
        symbol_map={
            False: "circle",
            True: "x"
        },
        color_discrete_map={
            "Adelie": "darkorange",
            "Chinstrap": "purple",
            "Gentoo": "#008B8B"
        },
        title="PCA Clusters vs Ground Truth"
    )
    return fig
```

```python
plot_pca_clusters(penguins_df)
```
